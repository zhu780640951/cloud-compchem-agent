### 查看 Colab 环境配置
运行以下代码来获取当前运行时的 CPU、内存、GPU 以及系统版本信息。

```python
import psutil
import platform
import subprocess

print("=== 操作系统信息 ===")
print(f"OS: {platform.system()} {platform.release()} ({platform.version()})")

print("\n=== CPU 信息 ===")
# CPU cores
physical_cores = psutil.cpu_count(logical=False)
total_cores = psutil.cpu_count(logical=True)
print(f"物理核心数: {physical_cores}")
print(f"逻辑核心数: {total_cores}")

# Try to get CPU model name
try:
    cpu_model = subprocess.check_output("cat /proc/cpuinfo | grep 'model name' | uniq", shell=True).decode().strip()
    print(cpu_model)
except Exception:
    pass

print("\n=== 内存信息 ===")
mem = psutil.virtual_memory()
total_gb = mem.total / (1024 ** 3)
available_gb = mem.available / (1024 ** 3)
print(f"总内存: {total_gb:.2f} GB")
print(f"可用内存: {available_gb:.2f} GB")

print("\n=== GPU 信息 ===")
try:
    gpu_info = subprocess.check_output("nvidia-smi", shell=True).decode()
    print(gpu_info)
except Exception:
    print("当前环境未启用 GPU (如果需要，请在 '代码执行程序' -> '更改运行时类型' 中开启 GPU)")
```
