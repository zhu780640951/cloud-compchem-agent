```python
import os
import psutil
import subprocess
import platform

print("=" * 50)
print("Google Colab CPU 硬件检测")
print("=" * 50)

print("\n【1】CPU 详细信息")
subprocess.run("lscpu", shell=True)

print("\n【2】CPU 核心数")
print("系统可见逻辑核心:", os.cpu_count())
print("进程可用逻辑核心:",
      len(os.sched_getaffinity(0)))

print("\n【3】内存信息")
ram = psutil.virtual_memory()
print(f"总内存: {ram.total / 1024**3:.2f} GB")
print(f"可用内存: {ram.available / 1024**3:.2f} GB")

print("\n【4】CPU 配额")
subprocess.run(
    "cat /sys/fs/cgroup/cpu.max",
    shell=True
)

print("\n【5】CPU 架构")
print(platform.machine())

print("\n【6】磁盘")
disk = psutil.disk_usage("/")
print(f"总空间: {disk.total / 1024**3:.2f} GB")
print(f"剩余空间: {disk.free / 1024**3:.2f} GB")
```

```python
import numpy as np
import time
from threadpoolctl import threadpool_limits

# 创建两个矩阵
np.random.seed(42)
A = np.random.rand(2000, 2000)
B = np.random.rand(2000, 2000)

# 限制为 2 个 CPU 计算线程
with threadpool_limits(limits=2):
    _ = A @ B  # 预热
    start = time.perf_counter()
    C = A @ B
    elapsed = time.perf_counter() - start

# 计算性能
gflops = (2 * 2000**3) / elapsed / 1e9

print("===== CPU 性能测试 =====")
print(f"矩阵规模：2000 × 2000")
print(f"计算耗时：{elapsed:.3f} 秒")
print(f"估算性能：{gflops:.2f} GFLOPS")
```
